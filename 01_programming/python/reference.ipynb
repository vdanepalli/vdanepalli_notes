# <center> **Understanding `df.pivot()` in Pandas**

----

### Explanation: Basic Pivot - Single Index, Single Column, Single Value
- **`index='Date'`**: Rows are defined by unique dates.
- **`columns='Product'`**: Columns are created for each unique product.
- **`values='Sales'`**: The `Sales` values fill the cells.

### Explanation: Pivot with Multi-Index
- **`index=['Date', 'Product']`**: Both `Date` and `Product` are used together to form the index.
- **`columns=None`**: No new columns are created; the pivoted data is organized by multi-index.
- **`values='Sales'`**: The `Sales` values are retained.

### Explanation: Pivot Without Values (Entire Row Data)
- **No `values` specified**: The `Sales` data is pivoted across products, resulting in a multi-level column index.

### Explanation: Handling Duplicates with pivot_table
- **`aggfunc='sum'`**: When there are duplicates (e.g., multiple entries for a date-product combination), `pivot_table` handles it by aggregating with the specified function (`sum` in this case).

### Explanation: Pivot with Missing Data
- **Missing values filled with `NaN`**: If no data exists for a particular index-column combination, `NaN` is inserted.

### Explanation: Pivot with Multiple Values
- **`values=['Sales', 'Quantity']`**: Pivoting multiple columns results in a multi-level column index with each value type (Sales and Quantity).

----

### Explanation: Pivot with Multi-Index (Corrected)
- **`index=['Date', 'Product']`**: Both `Date` and `Product` are used together to form a multi-index.
- **`values='Sales'`**: The `Sales` values are aggregated based on the multi-index.
- **`aggfunc='sum'`**: Aggregates values using the sum function to handle duplicates.

```python
import pandas as pd

# Sample data
data = {
    'Date': ['2023-01', '2023-01', '2023-02', '2023-02', '2023-03', '2023-03'],
    'Product': ['A', 'B', 'A', 'B', 'A', 'B'],
    'Sales': [250, 150, 200, 100, 300, 180],
    'Quantity': [10, 15, 8, 7, 12, 14]
}
df = pd.DataFrame(data)

# Example 1: Basic Pivot - Single Index, Single Column, Single Value
pivot_basic = df.pivot(index='Date', columns='Product', values='Sales')

# Example 2: Pivot with Multi-Index
# pivot_multi_index = df.pivot(index=['Date', 'Product'], columns=None, values='Sales')

# Example 3: Pivot Without Values (Entire Row Data)
pivot_no_values = df.pivot(index='Date', columns='Product')

# Example 4: Handling Duplicates with pivot_table (Using aggregation)
# For this, we’ll intentionally duplicate data to show how pivot_table works
data_with_duplicates = {
    'Date': ['2023-01', '2023-01', '2023-02', '2023-02', '2023-03', '2023-03', '2023-01'],
    'Product': ['A', 'B', 'A', 'B', 'A', 'B', 'A'],
    'Sales': [250, 150, 200, 100, 300, 180, 250]
}
df_duplicates = pd.DataFrame(data_with_duplicates)
pivot_with_duplicates = df_duplicates.pivot_table(index='Date', columns='Product', values='Sales', aggfunc='sum')

# Example 5: Pivot with Missing Data
data_with_missing = {
    'Date': ['2023-01', '2023-01', '2023-02', '2023-03', '2023-03'],
    'Product': ['A', 'B', 'A', 'A', 'B'],
    'Sales': [250, 150, 200, 300, 180]
}
df_missing = pd.DataFrame(data_with_missing)
pivot_with_missing = df_missing.pivot(index='Date', columns='Product', values='Sales')

# Example 6: Pivot with Multiple Values
pivot_multi_values = df.pivot(index='Date', columns='Product', values=['Sales', 'Quantity'])

# Display all pivots
print("Basic Pivot:\n", pivot_basic, "\n")
# print("Pivot with Multi-Index:\n", pivot_multi_index, "\n")
print("Pivot Without Values:\n", pivot_no_values, "\n")
print("Pivot Handling Duplicates with pivot_table:\n", pivot_with_duplicates, "\n")
print("Pivot with Missing Data:\n", pivot_with_missing, "\n")
print("Pivot with Multiple Values:\n", pivot_multi_values, "\n")

# Corrected Example: Pivot with Multi-Index using pivot_table
pivot_multi_index_corrected = df.pivot_table(index=['Date', 'Product'], values='Sales', aggfunc='sum')
print("Pivot with Multi-Index (Corrected):\n", pivot_multi_index_corrected)
```

# <center> **Understanding `df.plot()` in Pandas**

`df.plot()` is a versatile plotting function in Pandas that allows for easy and quick visualization of DataFrame data. It integrates with Matplotlib to offer a wide range of plots. Below are examples demonstrating various uses and parameters of `df.plot()`.

---

### **Explanation of Examples**

#### **Example 1: Default Line Plot**
- **Description**: The `df.plot()` method with no parameters generates a line plot. This is the default plot type in Pandas. Each column is plotted as a separate line.
- **Use Case**: Ideal for time series data and any continuous data where you want to observe trends.

#### **Example 2: Bar Plot**
- **Description**: Setting `kind='bar'` changes the plot to a bar plot. Each column is represented as a group of bars.
- **Use Case**: Useful for comparing categories or discrete values.

#### **Example 3: Horizontal Bar Plot**
- **Description**: Similar to the bar plot, but bars are horizontal. Achieved by setting `kind='barh'`.
- **Use Case**: Better for displaying long category names or when vertical space is constrained.

#### **Example 4: Histogram**
- **Description**: Using `kind='hist'` creates a histogram, which is useful for displaying the distribution of a dataset.
- **Use Case**: Ideal for statistical analysis and understanding the frequency distribution of data.

#### **Example 5: Box Plot**
- **Description**: A box plot is created using `kind='box'`. It visualizes the distribution of data based on quartiles.
- **Use Case**: Useful for detecting outliers and understanding data dispersion.

#### **Example 6: Area Plot**
- **Description**: By setting `kind='area'`, an area plot is generated. It's similar to a line plot but filled under the lines.
- **Use Case**: Effective for showing cumulative totals over time or space.

#### **Example 7: Scatter Plot**
- **Description**: Using `kind='scatter'`, a scatter plot is created, plotting one variable against another (`x='A'`, `y='B'`).
- **Use Case**: Best for showing relationships between two continuous variables.

#### **Example 8: Hexbin Plot**
- **Description**: The hexbin plot (`kind='hexbin'`) is useful for visualizing the relationship between two variables for large datasets by using hexagonal bins.
- **Use Case**: Ideal for dense data where scatter plots may become overplotted.

#### **Example 9: Pie Plot**
- **Description**: `kind='pie'` creates a pie chart from a single column.
- **Use Case**: Suitable for displaying proportions of a whole, though generally not recommended for large datasets or many categories.

---

### **Additional Notes on `df.plot()`**

- **Customization**: Each plot can be customized using additional parameters like `title`, `xlabel`, `ylabel`, `legend`, `color`, etc. Matplotlib's functionalities can also be used for further customization.
- **Optimization Tips**:
  - Use vectorized operations when preparing data for plotting.
  - For large datasets, consider sampling or aggregating data before plotting to avoid performance issues.
  - Use Matplotlib directly for complex plots or when you need finer control over the plot elements.
- **Industry Standards**:
  - Data visualization is crucial for understanding and communicating data insights. Standard practices include:
    - Keeping plots simple and focused on conveying a clear message.
    - Using appropriate plot types based on data characteristics.
    - Ensuring plots are properly labeled with titles, axis labels, and legends.
    - Maintaining consistency in styling and colors for easier interpretation.
- **Common Use Cases**:
  - Exploratory Data Analysis (EDA): Quickly visualizing data to identify trends, patterns, and anomalies.
  - Reporting: Creating visual summaries of key metrics for stakeholders.
  - Dashboarding: Integrating plots into web dashboards for real-time data monitoring.

```python
# Import necessary libraries
import pandas as pd
import matplotlib.pyplot as plt

# Sample DataFrame
data = {
    'A': [1, 2, 3, 4, 5],
    'B': [10, 20, 30, 40, 50],
    'C': [5, 3, 6, 9, 12]
}
df = pd.DataFrame(data)

# Example 1: Default Line Plot
# Plotting the DataFrame as a line plot.
print("Example 1: Default Line Plot")
df.plot(title='Line Plot')
plt.show()

# Example 2: Bar Plot
# Plotting the DataFrame as a bar plot.
print("Example 2: Bar Plot")
df.plot(kind='bar', title='Bar Plot')
plt.show()

# Example 3: Horizontal Bar Plot
# Plotting the DataFrame as a horizontal bar plot.
print("Example 3: Horizontal Bar Plot")
df.plot(kind='barh', title='Horizontal Bar Plot')
plt.show()

# Example 4: Histogram
# Plotting the DataFrame as a histogram.
print("Example 4: Histogram")
df.plot(kind='hist', title='Histogram')
plt.show()

# Example 5: Box Plot
# Plotting the DataFrame as a box plot.
print("Example 5: Box Plot")
df.plot(kind='box', title='Box Plot')
plt.show()

# Example 6: Area Plot
# Plotting the DataFrame as an area plot.
print("Example 6: Area Plot")
df.plot(kind='area', title='Area Plot')
plt.show()

# Example 7: Scatter Plot
# Plotting a scatter plot using columns 'A' and 'B'.
print("Example 7: Scatter Plot")
df.plot(kind='scatter', x='A', y='B', title='Scatter Plot')
plt.show()

# Example 8: Hexbin Plot
# Plotting a hexbin plot using columns 'A' and 'B'.
print("Example 8: Hexbin Plot")
df.plot(kind='hexbin', x='A', y='B', gridsize=25, title='Hexbin Plot')
plt.show()

# Example 9: Pie Plot
# Plotting a pie plot using column 'B'.
print("Example 9: Pie Plot")
df.plot(kind='pie', y='B', autopct='%1.1f%%', title='Pie Plot')
plt.show()
```
